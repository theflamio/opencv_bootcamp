# Annotating Images i Computer Vision

I computer vision refererer **annotating images** til processen, hvor billeder forsynes med **labels og strukturerede annotationer**, som anvendes til træning og evaluering af maskinlæringsmodeller.

Annotering er en forudsætning for **supervised learning**, da modellen lærer sammenhænge mellem billeddata og de tilknyttede labels.

---

## Formål med annotering
Formålet med annotering er at:
- Give semantisk mening til rå billeddata
- Muliggøre læring af mønstre i visuelle data
- Understøtte træning, validering og test af modeller

---

## Typer af billedannotering

### 1. Klassifikation
- Hvert billede tildeles **én klasse-label**
- Der angives ikke, hvor objektet befinder sig i billedet

**Anvendelse**
- Bruges når billedet indeholder ét dominerende objekt

**Eksempler på labels**
- `tank`
- `truck`
- `civilian`

---

### 2. Object Detection
- Objekter lokaliseres ved hjælp af **bounding boxes**
- Hver bounding box tildeles en klasse-label

**Annotation består typisk af**
```
[x_min, y_min, x_max, y_max, class]
```

**Anvendelse**
- Bruges når både objektets type og position er relevante

---

### 3. Segmentation
- Hvert pixel i billedet tildeles en klasse
- Giver en præcis afgrænsning af objekters form

**Typer**
- Semantic segmentation
- Instance segmentation

**Anvendelse**
- Bruges når detaljeret geometrisk information er nødvendig

---

### 4. Keypoints / Landmarks
- Specifikke punkter på et objekt annoteres
- Anvendes til strukturel og geometrisk analyse

**Eksempler**
- Tårn
- Kanonløb
- Hjul

---

## Betydning af annoteringskvalitet
Annoteringens kvalitet har direkte indflydelse på modellens ydeevne.

- Inkorrekte eller inkonsistente labels reducerer modelpræcision
- Konsistente og præcise annotationer forbedrer generalisering

I praksis har annoteringskvalitet ofte større betydning end valg af modelarkitektur.

---

## Værktøjer til billedannotering
Eksempler på udbredte annoteringsværktøjer:

- **LabelImg** – bounding box-annotering
- **CVAT** – understøtter detection, segmentation og video
- **Roboflow** – webbaseret platform med dataset-håndtering
- **VGG Image Annotator (VIA)** – letvægts, browserbaseret værktøj

---

## Typisk workflow i et computer vision-projekt

1. Indsamling af billeddata
2. Annotering af billeder
3. Opdeling i trænings-, validerings- og testdatasæt
4. Træning af model (fx YOLO)
5. Evaluering og visualisering af resultater

---

## Opsummering
Annotering af billeder er en central proces i computer vision og danner grundlaget for al supervised læring.

Korrekt annoterede datasæt er afgørende for at opnå pålidelige og robuste modeller.
